```python
import pandas as pd
from pathlib import Path
from sklearn.preprocessing import StandardScaler
```

```python
# Cargar dataset
csv_path = Path().resolve().parent / "data" / "WA_Fn-UseC_-Telco-Customer-Churn.csv"
df = pd.read_csv(csv_path)
```

```python
df.head()
```

```python
# Eliminar CustumerID ya que no aporta ningun valor
df = df.drop(["customerID"], axis=1)
```

```python
# Convertir 'TotalCharges' a numérico
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
# Convertir 'SeniorCitizen' a categórico
df["SeniorCitizen"] = df["SeniorCitizen"].astype(str)
```

```python
# Aplicar One-Hot Encoding a variables categóricas
df = pd.get_dummies(df, drop_first=True)
```

```python
# Manejar valores nulos
df.fillna(df.median(), inplace=True)
```

```python
# Escalar variables numéricas
scaler = StandardScaler()
numeric_cols = ["MonthlyCharges", "TotalCharges", "tenure"]
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
```

```python
df.head()
```

```python
# Definir la ruta para guardar el dataset limpio
cleaned_csv_path = Path().resolve().parent / "data" / "cleaned_telco_churn.csv"
# Guardar dataset limpio
df.to_csv(cleaned_csv_path, index=False)
```
